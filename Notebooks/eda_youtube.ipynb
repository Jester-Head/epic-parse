```python
def get_avg_comments(df):
    avg_comments_per_video = df.groupby('channel_name')['text'].count()
    avg_comments_per_video = pd.DataFrame(avg_comments_per_video)
    avg_comments_per_video['video_count'] = df.groupby('channel_name')['video_id'].nunique()
    avg_comments_per_video['avg_comments'] = (avg_comments_per_video['text'] / avg_comments_per_video['video_count']).round(2)
    return avg_comments_per_video.sort_values(by='avg_comments', ascending=False)
```

```python
import pandas as pd
import datetime as dt

# Convert 'updated_at' to datetime
df['updated_at'] = pd.to_datetime(df['updated_at'])

# 1. Calculate total comments per channel
total_comments = df.groupby('channel_name')['text'].count().reset_index()
total_comments.columns = ['channel_name', 'total_comments']

# 2. Calculate recent comments (last 30 days)
cutoff_date = df['updated_at'].max() - dt.timedelta(days=30)
recent_df = df[df['updated_at'] >= cutoff_date]
recent_comments = recent_df.groupby('channel_name')['text'].count().reset_index()
recent_comments.columns = ['channel_name', 'recent_comments']

# 3. Merge both DataFrames
merged = pd.merge(total_comments, recent_comments, on='channel_name', how='left').fillna(0)

# 4. Sort by total_comments or recent_comments
merged = merged.sort_values(by=['total_comments', 'recent_comments'], ascending=False)
```

```python
import math

def n_margin_of_error(p=0.5, e=0.10, alpha=0.05):
    z = 1.96 if alpha == 0.05 else 1.645  # tweak if needed
    return math.ceil((z**2 * p * (1 - p)) / (e**2))

def n_one_hit(p_jar=0.05, guarantee=0.95):
    return math.ceil(math.log(1 - guarantee) / math.log(1 - p_jar))

print("Margin-error 95 %, ±10 pp :", n_margin_of_error())
print("≥1 jargon hit 95 %, p=5 % :", n_one_hit())
```

```python
import pandas as pd

def sample_comments_by_channel(
    df: pd.DataFrame,
    channel_col: str = "channel_id",
    k_min: int = 50,
    n_total: int = 10_000,
    random_state: int = 42,
):
    """
    Stratified sampler that guarantees at least `k_min` rows per channel,
    then fills the remaining budget proportionally to channel size.

    Parameters
    ----------
    df : pd.DataFrame
        Must contain a column with channel IDs and the comment text.
    channel_col : str, default "channel_id"
        Column name that identifies the channel for each row.
    k_min : int, default 50
        Minimum number of rows to pull from every channel (or all rows if the
        channel has < k_min total).
    n_total : int, default 10_000
        Total number of rows you want in the final sample.
    random_state : int, default 42
        Seed for reproducibility.

    Returns
    -------
    pd.DataFrame
        Sampled dataframe, shuffled.
    """
    rng = random_state

    # --- 1. “Floor” draw ----------------------------------------------------
    floor_samples = (
        df.groupby(channel_col, group_keys=False)
          .apply(lambda g: g.sample(min(len(g), k_min), random_state=rng))
    )

    # --- 2. Proportional top-up --------------------------------------------
    remaining_budget = n_total - len(floor_samples)
    if remaining_budget > 0:
        leftovers = df.drop(floor_samples.index)
        if not leftovers.empty:
            frac = min(1.0, remaining_budget / len(leftovers))
            extra_samples = (
                leftovers.groupby(channel_col, group_keys=False)
                         .apply(lambda g: g.sample(frac=frac, random_state=rng))
            )
            sampled_df = pd.concat([floor_samples, extra_samples])
        else:
            sampled_df = floor_samples
    else:
        # Already filled the budget with the floor samples alone
        sampled_df = floor_samples.sample(n_total, random_state=rng)

    # --- 3. Final shuffle ---------------------------------------------------
    return sampled_df.sample(frac=1.0, random_state=rng).reset_index(drop=True)
```
